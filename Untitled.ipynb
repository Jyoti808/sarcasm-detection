```python
# Import library for text preprocessing

import re


# Function to clean headlines

def clean_text(text):
    
    # Convert text to lowercase
    text = text.lower()
    
    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)
    
    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)
    
    # Keep letters, numbers, apostrophes and spaces
    text = re.sub(r"[^a-z0-9'\s]", " ", text)
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    return text


print("Text cleaning function created successfully.")
```

```python
# Apply text cleaning function to headlines

df["cleaned_headline"] = df["headline"].apply(clean_text)

print("Text cleaning completed successfully.")

# Compare original and cleaned headlines
df[["headline", "cleaned_headline", "is_sarcastic"]].head(10)
```

```python
# Check cleaned headline quality

print("Total Headlines:", len(df))

print(
    "Empty Cleaned Headlines:",
    (df["cleaned_headline"].str.strip() == "").sum()
)

print(
    "Missing Cleaned Headlines:",
    df["cleaned_headline"].isnull().sum()
)

print("\nCleaned Headline Length Statistics:")

df["cleaned_word_count"] = (
    df["cleaned_headline"]
    .apply(lambda x: len(x.split()))
)

print(df["cleaned_word_count"].describe())
```

```python
# Import train-test split

from sklearn.model_selection import train_test_split


# Define features and target

X = df["cleaned_headline"]
y = df["is_sarcastic"]


# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# Display dataset split information

print("Training Samples:", len(X_train))
print("Testing Samples :", len(X_test))

print("\nTraining Class Distribution:")
print(y_train.value_counts())

print("\nTesting Class Distribution:")
print(y_test.value_counts())
```

```python
# Import TF-IDF Vectorizer

from sklearn.feature_extraction.text import TfidfVectorizer


# Create TF-IDF vectorizer

tfidf = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2),
    min_df=2
)


# Fit on training data and transform both datasets

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)


# Display TF-IDF information

print("TF-IDF Vectorization Completed.")

print("\nTraining TF-IDF Shape:", X_train_tfidf.shape)
print("Testing TF-IDF Shape :", X_test_tfidf.shape)

print("\nVocabulary Size:", len(tfidf.vocabulary_))
```

```python
# Import Logistic Regression

from sklearn.linear_model import LogisticRegression


# Create Logistic Regression model

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)


# Train the model

lr_model.fit(X_train_tfidf, y_train)

print("Logistic Regression model trained successfully.")
```

```python
# Import evaluation metrics

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)


# Make predictions

y_pred_lr = lr_model.predict(X_test_tfidf)

# Probability predictions for ROC-AUC
y_prob_lr = lr_model.predict_proba(X_test_tfidf)[:, 1]


# Calculate evaluation metrics

lr_accuracy = accuracy_score(y_test, y_pred_lr)
lr_precision = precision_score(y_test, y_pred_lr)
lr_recall = recall_score(y_test, y_pred_lr)
lr_f1 = f1_score(y_test, y_pred_lr)
lr_roc_auc = roc_auc_score(y_test, y_prob_lr)


# Display results

print("Logistic Regression Evaluation")
print("-" * 45)

print(f"Accuracy  : {lr_accuracy:.4f}")
print(f"Precision : {lr_precision:.4f}")
print(f"Recall    : {lr_recall:.4f}")
print(f"F1 Score  : {lr_f1:.4f}")
print(f"ROC-AUC   : {lr_roc_auc:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_lr,
        target_names=["Non-Sarcastic", "Sarcastic"]
    )
)
```

```python
# Confusion Matrix for Logistic Regression

from sklearn.metrics import confusion_matrix

cm_lr = confusion_matrix(y_test, y_pred_lr)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Non-Sarcastic", "Sarcastic"],
    yticklabels=["Non-Sarcastic", "Sarcastic"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.show()
```

```python
# Check TensorFlow and Keras installation

import tensorflow as tf
import keras

print("TensorFlow Version:", tf.__version__)
print("Keras Version:", keras.__version__)

print("\nTensorFlow and Keras imported successfully.")
```

```python
# Import Tokenizer

from tensorflow.keras.preprocessing.text import Tokenizer


# Set maximum vocabulary size

vocab_size = 20000


# Create tokenizer

tokenizer = Tokenizer(
    num_words=vocab_size,
    oov_token="<OOV>"
)


# Fit tokenizer ONLY on training headlines

tokenizer.fit_on_texts(X_train)


# Get complete vocabulary information

total_words = len(tokenizer.word_index) + 1

print("Tokenizer fitted successfully.")

print("\nTotal Unique Words in Training Data:", total_words)
print("Vocabulary Size Used by Model:", vocab_size)

print("\nFirst 10 Words in Vocabulary:")

for word, index in list(tokenizer.word_index.items())[:10]:
    print(word, ":", index)
```

```python
# Convert training and testing headlines into sequences

X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)


# Calculate sequence lengths

train_seq_lengths = [len(sequence) for sequence in X_train_seq]


# Display sequence length statistics

print("Text converted to sequences successfully.")

print("\nTraining Sequences:", len(X_train_seq))
print("Testing Sequences :", len(X_test_seq))

print("\nSequence Length Statistics:")
print("Minimum Length :", np.min(train_seq_lengths))
print("Mean Length    :", np.mean(train_seq_lengths))
print("Median Length  :", np.median(train_seq_lengths))
print("Maximum Length :", np.max(train_seq_lengths))

print("\nSequence Length Percentiles:")
print("90th Percentile:", np.percentile(train_seq_lengths, 90))
print("95th Percentile:", np.percentile(train_seq_lengths, 95))
print("99th Percentile:", np.percentile(train_seq_lengths, 99))
```

```python
# Import padding function

from tensorflow.keras.preprocessing.sequence import pad_sequences


# Set maximum sequence length

max_length = 20


# Pad training sequences

X_train_pad = pad_sequences(
    X_train_seq,
    maxlen=max_length,
    padding="post",
    truncating="post"
)


# Pad testing sequences

X_test_pad = pad_sequences(
    X_test_seq,
    maxlen=max_length,
    padding="post",
    truncating="post"
)


# Display padded sequence information

print("Sequence padding completed successfully.")

print("\nTraining Data Shape:", X_train_pad.shape)
print("Testing Data Shape :", X_test_pad.shape)

print("\nExample Original Headline:")
print(X_train.iloc[0])

print("\nExample Token Sequence:")
print(X_train_seq[0])

print("\nExample Padded Sequence:")
print(X_train_pad[0])
```

```python
# Import layers and Sequential model

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding,
    LSTM,
    Dense,
    Dropout
)


# Set embedding dimension

embedding_dim = 100


# Build LSTM model

lstm_model = Sequential([
    
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        input_length=max_length
    ),
    
    LSTM(
        64,
        dropout=0.2,
        recurrent_dropout=0.2
    ),
    
    Dense(
        32,
        activation="relu"
    ),
    
    Dropout(0.3),
    
    Dense(
        1,
        activation="sigmoid"
    )
])


# Compile model

lstm_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# Build model to display parameter shapes

lstm_model.build(
    input_shape=(None, max_length)
)


# Display architecture

lstm_model.summary()
```

```python
# Import callbacks

from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint


# Early stopping

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True,
    verbose=1
)


# Save the best LSTM model

model_checkpoint = ModelCheckpoint(
    "best_lstm_model.keras",
    monitor="val_loss",
    save_best_only=True,
    verbose=1
)


print("Callbacks created successfully.")
```

```python
# Train LSTM model

history_lstm = lstm_model.fit(
    X_train_pad,
    y_train,
    epochs=15,
    batch_size=64,
    validation_split=0.20,
    callbacks=[
        early_stopping,
        model_checkpoint
    ],
    verbose=1
)
```

```python
# Plot LSTM training and validation performance

epochs_range = range(1, len(history_lstm.history["accuracy"]) + 1)

plt.figure(figsize=(8, 5))

plt.plot(
    epochs_range,
    history_lstm.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    epochs_range,
    history_lstm.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.title("LSTM Training vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Plot LSTM training and validation loss

epochs_range = range(1, len(history_lstm.history["loss"]) + 1)

plt.figure(figsize=(8, 5))

plt.plot(
    epochs_range,
    history_lstm.history["loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    epochs_range,
    history_lstm.history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.title("LSTM Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Evaluate LSTM model on test data

# Predict probabilities
y_prob_lstm = lstm_model.predict(
    X_test_pad,
    verbose=0
).ravel()

# Convert probabilities to class labels
y_pred_lstm = (y_prob_lstm >= 0.5).astype(int)


# Calculate evaluation metrics

lstm_accuracy = accuracy_score(y_test, y_pred_lstm)
lstm_precision = precision_score(y_test, y_pred_lstm)
lstm_recall = recall_score(y_test, y_pred_lstm)
lstm_f1 = f1_score(y_test, y_pred_lstm)
lstm_roc_auc = roc_auc_score(y_test, y_prob_lstm)


# Display results

print("LSTM Evaluation")
print("-" * 45)

print(f"Accuracy  : {lstm_accuracy:.4f}")
print(f"Precision : {lstm_precision:.4f}")
print(f"Recall    : {lstm_recall:.4f}")
print(f"F1 Score  : {lstm_f1:.4f}")
print(f"ROC-AUC   : {lstm_roc_auc:.4f}")

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred_lstm,
        target_names=["Non-Sarcastic", "Sarcastic"]
    )
)
```

```python
# Confusion Matrix for LSTM

cm_lstm = confusion_matrix(y_test, y_pred_lstm)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_lstm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Non-Sarcastic", "Sarcastic"],
    yticklabels=["Non-Sarcastic", "Sarcastic"]
)

plt.title("Confusion Matrix - LSTM")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.show()
```

```python
# Function to predict sarcasm from a new headline

from tensorflow.keras.preprocessing.sequence import pad_sequences

def predict_sarcasm(headline):

    # Clean input text
    cleaned_text = clean_text(headline)

    # Convert text to sequence
    sequence = tokenizer.texts_to_sequences([cleaned_text])

    # Pad sequence
    padded_sequence = pad_sequences(
        sequence,
        maxlen=max_length,
        padding="post",
        truncating="post"
    )

    # Predict using final LSTM model
    probability = lstm_model.predict(
        padded_sequence,
        verbose=0
    )[0][0]

    # Determine predicted class
    if probability >= 0.5:
        prediction = "Sarcastic"
    else:
        prediction = "Non-Sarcastic"

    # Display result
    print("Headline   :", headline)
    print("Prediction :", prediction)
    print(f"Sarcasm Probability : {probability:.4f}")

    return prediction, probability
```

```python
%%writefile app.py

import streamlit as st
import tensorflow as tf
import pickle
import json
import re
from tensorflow.keras.preprocessing.sequence import pad_sequences


# Page configuration
st.set_page_config(
    page_title="Sarcasm Detection",
    page_icon="😏",
    layout="centered"
)


# Text cleaning function
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"<.*?>", "", text)
    text = re.sub(r"[^a-z0-9'\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text


# Load trained model
@st.cache_resource
def load_sarcasm_model():
    return tf.keras.models.load_model(
        "final_sarcasm_lstm_model.keras"
    )


# Load tokenizer
@st.cache_resource
def load_tokenizer():
    with open("sarcasm_tokenizer.pkl", "rb") as file:
        return pickle.load(file)


# Load configuration
with open("sarcasm_model_config.json", "r") as file:
    config = json.load(file)


model = load_sarcasm_model()
tokenizer = load_tokenizer()

max_length = config["max_length"]
threshold = config["classification_threshold"]


# App title
st.title("😏 Sarcasm Detection System")

st.write(
    "Enter a news headline or sentence below to detect "
    "whether it is sarcastic or non-sarcastic."
)


# User input
user_text = st.text_area(
    "Enter Headline:",
    placeholder="Example: Absolutely love waiting three hours at the airport..."
)


# Prediction button
if st.button("Detect Sarcasm"):

    if user_text.strip():

        # Clean input
        cleaned_text = clean_text(user_text)

        # Convert text to sequence
        sequence = tokenizer.texts_to_sequences([cleaned_text])

        # Pad sequence
        padded_sequence = pad_sequences(
            sequence,
            maxlen=max_length,
            padding="post",
            truncating="post"
        )

        # Predict
        probability = model.predict(
            padded_sequence,
            verbose=0
        )[0][0]

        # Display prediction
        if probability >= threshold:
            st.error("😏 Prediction: Sarcastic")
        else:
            st.success("🙂 Prediction: Non-Sarcastic")

        st.write(
            f"**Sarcasm Probability:** {probability * 100:.2f}%"
        )

    else:
        st.warning("Please enter a headline first.")
```
